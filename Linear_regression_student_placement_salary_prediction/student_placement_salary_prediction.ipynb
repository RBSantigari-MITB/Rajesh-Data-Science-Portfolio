```python
df = df.drop(['student_id','branch'],axis = 1)
df
```

```python
df = df[df['salary_lpa']>0]
```

```python
df.shape
```

```python
df_encoded = pd.get_dummies(df,columns = ['company_type','job_role'],drop_first = True)
```

```python
df_encoded.info()
```

MODEL TRAINING 

```python
X=df_encoded.drop('salary_lpa',axis=1)
```

```python
Y=df['salary_lpa']
```

```python
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train
```

```python
Y_test
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model.fit(X_train,Y_train)
```

```python
Y_pred = model.predict(X_test)
```

```python
comparision = pd.DataFrame({'Actual':Y_test,'Predicted':Y_pred})
```

```python
comparision.head()
```

EVALUATION OF THE MODEL

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
mse = mean_squared_error(Y_test,Y_pred)
r2 = r2_score(Y_test,Y_pred)
```

```python
print(mse)
print(r2)
```

```python
rmse=np.sqrt(mse)
print(rmse)
print(r2)
```

CONCLUSION: The model achieved an R² score of 84.42%, indicating that it explains a significant portion of salary variation effectively.
The RMSE value of 6.57 LPA suggests that the model’s salary predictions deviate from actual salaries by approximately 6.57 LPA on average, which is reasonable considering the wide salary range present in the dataset.
